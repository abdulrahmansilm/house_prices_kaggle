```python
y_train= train['SalePrice']
train = train.drop(columns=['SalePrice'])
```

```python
from sklearn.tree import DecisionTreeRegressor

model = DecisionTreeRegressor(random_state=42)
model.fit(train, y_train)

print ("Training abgeschlossen")
```

```python
from sklearn.model_selection import cross_val_score

score = cross_val_score(model, train, y_train,cv=5)
print ("Genauigkeit: ", score.mean().round(3))
```

```python
from sklearn.ensemble import RandomForestRegressor

model_rf = RandomForestRegressor(n_estimators=100, random_state=42)
model_rf.fit(train, y_train)

score_rf= cross_val_score(model_rf, train,y_train, cv=5)
print ("Randim Forest: ", score_rf.mean().round(3))
```

```python
x_test = test.drop(columns=['Id'])
x_test = x_test.reindex(columns=train.columns, fill_value=0)


predictions=model_rf.predict(x_test)

submission=pd.DataFrame({
    'Id': test['Id'],
    'SalePrice' : predictions
})

submission.to_csv('submission.csv', index=False)
print(submission.head())
```
